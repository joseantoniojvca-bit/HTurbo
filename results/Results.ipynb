```python
import numpy as np
import ROOT
```

```python
f1 = ROOT.TFile.Open("born_SigInt.root")
f2 = ROOT.TFile.Open("born_Sig.root")
f3 = ROOT.TFile.Open("born_Int.root")
f1.ls()
```

```python
# Get histogram from file
h1 = f1.Get("s_m")   # replace with actual name from f.ls()
h2 = f2.Get("s_m")
h3 = f3.Get("s_m")

# --- Disable the stats box globally ---
ROOT.gStyle.SetOptStat(0)      # removes the stats box
h1.SetTitle("Cross section at Born level")
h1.GetXaxis().SetTitle('M_{H}[GeV]')
h1.GetYaxis().SetTitle("#sigma_{0}(gg #rightarrow H) [nb]")

h1.GetXaxis().SetRangeUser(124.94, 125.05)   # X range
h1.GetYaxis().SetRangeUser(-0.1*10**(-6), 1.3*10**(-6))   # Y range

h1.SetLineColor(ROOT.kRed)       # line color
h2.SetLineColor(ROOT.kBlue)
h3.SetLineColor(ROOT.kGreen+2)
#h.SetLineWidth(2)               # line thickness
h1.SetLineStyle(1)               # 1=solid, 2=dashed, 3=dotted, 4=dash-dot
h2.SetLineStyle(1)
h3.SetLineStyle(1)

#h.SetFillColor(ROOT.kBlue - 9)  # fill color
#h.SetFillStyle(1001)            # 1001=solid, 3004=hatched, 0=hollow

#ROOT.kBlack, ROOT.kRed, ROOT.kBlue, ROOT.kGreen
#ROOT.kMagenta, ROOT.kCyan, ROOT.kOrange, ROOT.kViolet

# Lighter/darker shades by adding/subtracting numbers
#ROOT.kRed + 2    # darker red
#ROOT.kBlue - 9   # lighter blue

h1.SetMarkerStyle(20)    # 20=circle, 21=square, 22=triangle, 33=diamond
h2.SetMarkerStyle(21)
h3.SetMarkerStyle(22)
h1.SetMarkerColor(ROOT.kBlack)
h2.SetMarkerColor(ROOT.kBlack)
h3.SetMarkerColor(ROOT.kBlack)
h1.SetMarkerSize(1)
h2.SetMarkerSize(1)
h3.SetMarkerSize(1)


# Font size
#h.GetXaxis().SetTitleSize(0.05)
#h.GetYaxis().SetLabelSize(0.04)

# Draw it
c = ROOT.TCanvas("c", "BORN", 2400, 1800)
h1.Draw("CP") # L: lines, C: smooth curve, P: points
h2.Draw("CP SAME")
h3.Draw("CP SAME")

leg = ROOT.TLegend(0.7, 0.7, 0.9, 0.9)   # (x1, y1, x2, y2) in pad coordinates
leg.AddEntry(h1, "BORN: SIG + INT", "lp")       # "l"=line, "f"=fill, "p"=marker
leg.AddEntry(h2, "BORN: SIG", "lp")
leg.AddEntry(h3, "BORN: INT", "lp")
leg.SetTextSize(25)
leg.Draw()

ymin, ymax = 0.8e-6, 1.25e-6
x_lo, x_hi = 124.995, 125.005
box = ROOT.TBox(x_lo, ymin, x_hi, ymax)
box.SetFillColorAlpha(ROOT.kGray, 0.3)
box.SetLineColor(ROOT.kGray+2)
box.SetLineStyle(2)
box.Draw("l")   # outline only

# --- Inset pad for the zoom ---
zoom_pad = ROOT.TPad("zoom_pad", "zoom", 0.12, 0.65, 0.4, 0.9)  # (x1,y1,x2,y2) in canvas fractions
zoom_pad.SetFillColor(0)
zoom_pad.SetBorderMode(0)
zoom_pad.Draw()
zoom_pad.SetGrid()
zoom_pad.cd()

# Clone histograms so the zoomed copies don't disturb the originals
h1z = h1.Clone("h1z")
h2z = h2.Clone("h2z")
h3z = h3.Clone("h3z")

h1z.GetXaxis().SetRangeUser(124.995, 125.005)   # the narrower region you want to zoom into
h1z.GetYaxis().SetRangeUser(0.8*10**(-6), 1.25*10**(-6))   # the narrower region you want to zoom into
h1z.SetTitle("")
h1z.Draw("CP")
h2z.Draw("CP SAME")
h3z.Draw("CP SAME")

# Optional: set higher resolution for fonts/lines
ROOT.gStyle.SetLineWidth(2)
c.SetGrid()
c.Draw()
```
